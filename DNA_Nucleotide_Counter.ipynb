```python
with open("sequence.fasta", "r") as file:
    lines = file.readlines()

dna = "".join(lines[1:]).replace("\n", "")
```

```python
print("Sequence length:", len(dna))
```

```python
count_A = dna.count("A")
count_T = dna.count("T")
count_G = dna.count("G")
count_C = dna.count("C")

print("A:", count_A)
print("T:", count_T)
print("G:", count_G)
print("C:", count_C)
```

```python
gc = (count_G + count_C) / len(dna) * 100

print(f"GC Content: {gc:.2f}%")
```

```python
counts = {base: dna.count(base) for base in "ATGC"}

print(counts)
```

```python
import matplotlib.pyplot as plt

bases = list(counts.keys())
values = list(counts.values())

plt.bar(bases, values)
plt.xlabel("Nucleotide")
plt.ylabel("Count")
plt.title("DNA Nucleotide Counts")
plt.show()
```
