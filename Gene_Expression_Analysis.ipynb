```python
import pandas as pd

file = "GSE5057_series_matrix.txt.gz"

df = pd.read_csv(
    file,
    sep="\t",
    comment="!",
    compression="gzip"
)

df.head()
```

```python
# Calculate the average expression across all samples for each gene
df["mean_expression"] = df.mean(axis=1)

# Display the first 5 genes with their mean expression
df.head()
```

```python
# Find the 10 genes with the highest mean expression
top10 = df["mean_expression"].sort_values(ascending=False).head(10)

print(top10)
```

```python
import matplotlib.pyplot as plt

# Plot the top 10 expressed genes
top10.sort_values().plot(kind="barh")

plt.title("Top 10 Expressed Genes")
plt.xlabel("Mean Expression")
plt.ylabel("Gene / Probe ID")
plt.tight_layout()

plt.savefig("top_genes.png")
plt.show()
```

## Gene Expression Analysis

Gene expression is the process by which information from a gene is used to produce a functional product, such as a protein. Gene expression values indicate the level of activity or abundance of a gene's transcript in a sample.

In this analysis, the mean expression value was calculated for each probe across the 11 samples in the GSE5057 dataset. The 10 probes with the highest mean expression values were then identified and visualized using a horizontal bar chart.

The highest-ranked probe was **212869_x_at**, with a mean expression value of approximately **33,707.13**. A high expression value indicates that the corresponding transcript was detected at a relatively high level across the analyzed samples. The high ranking may be biologically significant because highly expressed genes can represent genes that are particularly active or abundant in the tissue being studied.

The analysis demonstrates how Python and pandas can be used to process gene expression data, calculate average expression levels, identify highly expressed genes, and visualize the results.
