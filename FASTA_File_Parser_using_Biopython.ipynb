```python
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction
```

```python
records = list(SeqIO.parse('insulin.fasta', 'fasta'))
```

```python
for r in records:
    print(f'ID: {r.id}')
    print(f'Length: {len(r.seq)}')
    print(f'GC: {gc_fraction(r.seq)*100:.2f}%')
```

```python
for r in records:
    print(f'Sequence: {str(r.seq[:50])}...')
```

```python
import pandas as pd

summary = []

for r in records:
    summary.append({
        'ID': r.id,
        'Length': len(r.seq),
        'GC Content (%)': round(gc_fraction(r.seq) * 100, 2)
    })

df = pd.DataFrame(summary)
df
```
